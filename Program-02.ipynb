```python
#Code From ChatGPT
# Exploratory Data Analysis (EDA) on Students Dataset

import pandas as pd

# 1. Load the dataset
df = pd.read_csv("students.csv")

# 2. Display the complete dataset
print("========== STUDENTS DATASET ==========")
display(df)

# 3. Display first 5 records
print("\n========== FIRST 5 RECORDS ==========")
display(df.head())

# 4. Display last 5 records
print("\n========== LAST 5 RECORDS ==========")
display(df.tail())

# 5. Display dataset information
print("\n========== DATASET INFORMATION ==========")
df.info()

# 6. Display number of rows and columns
print("\n========== DATASET SHAPE ==========")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# 7. Display column names
print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

# 8. Display descriptive statistics
print("\n========== DESCRIPTIVE STATISTICS ==========")
display(df.describe())

# 9. Display descriptive statistics for all columns
print("\n========== STATISTICS FOR ALL COLUMNS ==========")
display(df.describe(include="all"))

# 10. Check missing values
print("\n========== MISSING VALUES ==========")
display(df.isnull().sum())

# 11. Calculate total missing values
print("\n========== TOTAL MISSING VALUES ==========")
total_missing = df.isnull().sum().sum()
print("Total missing values:", total_missing)

# 12. Calculate percentage of missing values
print("\n========== MISSING VALUE PERCENTAGE ==========")
missing_percentage = (df.isnull().sum() / len(df)) * 100
display(missing_percentage)

# 13. Check duplicate records
print("\n========== DUPLICATE RECORDS ==========")
duplicate_count = df.duplicated().sum()
print("Number of duplicate records:", duplicate_count)

# 14. Display duplicate records
print("\n========== DUPLICATE ROWS ==========")
display(df[df.duplicated()])

# 15. Display data types
print("\n========== DATA TYPES ==========")
display(df.dtypes)

# 16. Display number of unique values
print("\n========== UNIQUE VALUES COUNT ==========")
display(df.nunique())

# 17. Display unique values of each column
print("\n========== UNIQUE VALUES ==========")
for column in df.columns:
    print("\n", column)
    print(df[column].unique())

print("\n========== EDA COMPLETED ==========")
```
